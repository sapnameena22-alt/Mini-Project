```python
import os
import shutil
from pathlib import Path

# This line will automatically find the path of the 'Downloads' folder on any Windows computer
folder_path = r"C:\Users\edune\Downloads"
directory = Path(folder_path)

TRACKED_EXTENSIONS = {
    'Images': ['.jpg', '.jpeg', '.png', '.gif', '.bmp', '.svg'],
    'Documents': ['.pdf', '.docx', '.doc', '.txt', '.xlsx', '.pptx', '.csv'],
    'Videos': ['.mp4', '.mkv', '.avi', '.mov', '.flv'],
    'Music': ['.mp3', '.wav', '.aac'],
    'Setups': ['.exe', '.msi', '.dmg', '.pkg'],
    'Archives': ['.zip', '.rar', '.7z', '.tar', '.gz']
}

summary = {category: 0 for category in TRACKED_EXTENSIONS.keys()}
summary['Others'] = 0

if directory.exists():
    for item in directory.iterdir():
        if item.is_file():
            file_extension = item.suffix.lower()
            moved = False
            
            for category, extensions in TRACKED_EXTENSIONS.items():
                if file_extension in extensions:
                    dest_folder = directory / category
                    dest_folder.mkdir(exist_ok=True)
                    shutil.move(str(item), str(dest_folder / item.name))
                    summary[category] += 1
                    moved = True
                    break
                    
            if not moved:
                dest_folder = directory / 'Others'
                dest_folder.mkdir(exist_ok=True)
                shutil.move(str(item), str(dest_folder / item.name))
                summary['Others'] += 1
                
    print("=== Process Completed! ===")
    print(summary)
else:
    print("Folder not found.")
```
